```python
 import pandas as pd

df = pd.read_csv("insurance.csv")

print(df.head(10))
```

```python
print("Shape of Dataset:")
print(df.shape)

print("\nDataset Information:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())
```

```python
print("Null Values:")
print(df.isnull().sum())
```

```python
numerical = df.select_dtypes(include=['int64', 'float64']).columns
categorical = df.select_dtypes(include=['object']).columns

print("Numerical Columns:")
print(numerical)

print("\nCategorical Columns:")
print(categorical)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

num_cols = ['age', 'bmi', 'children', 'charges']

for col in num_cols:
    plt.figure(figsize=(6,4))
    sns.histplot(df[col], kde=True)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

```python
cat_cols = ['sex', 'smoker', 'region']

for col in cat_cols:
    plt.figure(figsize=(6,4))
    sns.countplot(x=col, data=df)
    plt.title(f"Count Plot of {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.show()
```

```python
print("Average Charges:", df['charges'].mean())
print("Maximum Charges:", df['charges'].max())
print("Minimum Charges:", df['charges'].min())

print("\nAverage Charges by Smoker:")
print(df.groupby('smoker')['charges'].mean())
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='smoker', y='charges', data=df)
plt.title("Charges vs Smoker")
plt.show()

plt.figure(figsize=(6,4))
sns.boxplot(x='sex', y='charges', data=df)
plt.title("Charges vs Sex")
plt.show()
```

```python
print("Average Age:", df['age'].mean())
print("Average BMI:", df['bmi'].mean())

print("\nAverage Charges by Smoker:")
print(df.groupby('smoker')['charges'].mean())

print("\nCustomers by Region:")
print(df['region'].value_counts())

# Observations:
# 1. Average age and BMI are displayed above.
# 2. Smokers usually have much higher insurance charges.
# 3. Region with maximum customers is shown above.
# 4. Heatmap and plots help identify relationships between variables.
```

```python
plt.figure(figsize=(8,6))

corr = df[['age','bmi','children','charges']].corr()

sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

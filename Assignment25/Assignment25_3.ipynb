```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

# Load Dataset
iris = load_iris()
X = iris.data
y = iris.target

# Split Dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33, random_state=42
)

# Different K Values
k_values = [3, 5, 7, 11, 13, 15]

best_score = 0
best_k = 0

print("KNN Manual Hyperparameter Tuning\n")

for k in k_values:
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train, y_train)

    score = model.score(X_test, y_test)
    print("n_neighbors =", k, " Accuracy =", round(score, 4))

    if score > best_score:
        best_score = score
        best_k = k

print("\nBest n_neighbors :", best_k)
print("Highest Accuracy :", round(best_score, 4))
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC

# Load Dataset
iris = load_iris()
X = iris.data
y = iris.target

# Split Dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33, random_state=42
)

# Hyperparameters
C_values = [1, 10, 20]
kernels = ["linear", "rbf"]

best_score = 0
best_C = 0
best_kernel = ""

print("Manual Hyperparameter Tuning - SVM\n")

for c in C_values:
    for k in kernels:
        model = SVC(C=c, kernel=k)
        model.fit(X_train, y_train)

        score = model.score(X_test, y_test)

        print("C =", c,
              " Kernel =", k,
              " Accuracy =", round(score, 4))

        if score > best_score:
            best_score = score
            best_C = c
            best_kernel = k

print("\nBest Hyperparameters")
print("C =", best_C)
print("Kernel =", best_kernel)
print("Best Accuracy =", round(best_score, 4))
```

```python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC

# Load Dataset
iris = load_iris()
X = iris.data
y = iris.target

# Split Dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33, random_state=42
)

# Parameter Grid
param_grid = {
    "C": [1, 10, 20],
    "kernel": ["linear", "rbf"]
}

# Grid Search
grid = GridSearchCV(SVC(), param_grid, cv=5)
grid.fit(X_train, y_train)

# Results
results = pd.DataFrame(grid.cv_results_)

print(results[[
    "param_C",
    "param_kernel",
    "mean_test_score"
]])

print("\nBest Parameters:")
print(grid.best_params_)

print("\nBest Score:")
print(round(grid.best_score_, 4))
```
